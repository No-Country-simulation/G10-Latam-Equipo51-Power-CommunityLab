```python
import json
from pathlib import Path

# Archivo de entrada con el array completo
INPUT_FILE = "entrada.json"

# Carpeta de salida
OUTPUT_DIR = Path("salida")
OUTPUT_DIR.mkdir(exist_ok=True)

# Cargar datos
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

contador = 1

for registro in data:
    origen = registro["origen_comunidad"]
    periodo = registro["periodo_referencia"]

    for interaccion in registro["interacciones"]:

        salida = {
            "origen_comunidad": origen,
            "periodo_referencia": periodo,
            "interaccion": interaccion
        }

        nombre_archivo = OUTPUT_DIR / f"mensaje_{contador:04d}.json"

        with open(nombre_archivo, "w", encoding="utf-8") as out:
            json.dump(
                salida,
                out,
                ensure_ascii=False,
                indent=2
            )

        contador += 1

print(f"Archivos generados: {contador - 1}")
```

```python
from pathlib import Path
from dotenv import load_dotenv
import os
import requests

# Cargar variables del .env
load_dotenv()

API_BUCKET = os.getenv("FAST_API_BUCKET")

if not API_BUCKET:
    raise ValueError(
        "No se encontró FAST_API_BUCKET en el archivo .env"
    )

# URLs construidas correctamente
API_URL = f"http://{API_BUCKET}:8000"
URL_UPLOAD = f"{API_URL}/upload"

print(f"API URL: {API_URL}")

# Carpeta donde están los JSON
DIRECTORIO_JSON = Path("salida")

if not DIRECTORIO_JSON.exists():
    raise FileNotFoundError(
        f"No existe la carpeta {DIRECTORIO_JSON}"
    )

exitosos = 0
errores = 0

for archivo in DIRECTORIO_JSON.glob("*.json"):

    try:

        with open(archivo, "rb") as f:

            files = {
                "file": (
                    archivo.name,
                    f,
                    "application/json"
                )
            }

            response = requests.post(
                URL_UPLOAD,
                files=files,
                timeout=30
            )

        if response.ok:
            exitosos += 1
            print(f"[OK] {archivo.name}")

        else:
            errores += 1
            print(
                f"[ERROR] {archivo.name} - "
                f"{response.status_code} - "
                f"{response.text}"
            )

    except Exception as e:
        errores += 1
        print(f"[ERROR] {archivo.name}: {e}")

print("\n===== RESUMEN =====")
print(f"Subidos: {exitosos}")
print(f"Errores: {errores}")
```
